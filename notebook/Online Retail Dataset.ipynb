```python
# Import libraries
import pandas as pd
```

```python
# Read excel file
df = pd.read_excel('Online Retail.xlsx')
```

```python
# Show shape of first 5 rows of the dataset
print(df.head())
```

```python
# Show data information
df.info()
```

```python
# Show missing values in the dataset
print(df.isnull().sum())
```

```python
# Drop null values from customer ID
df.dropna(subset=['CustomerID'], inplace=True)
```

```python
# Count of duplicate rows
print("\nDuplicate Rows")   
print(df.duplicated().sum())
```

```python
# Drop duplicate rows
df.drop_duplicates(inplace=True)
```

```python
#Print rows where Quantity or UnitPrice are zero or negative.
print("\nRows with Zero or Negative Quantity:", (df["Quantity"] <= 0).sum())
print("Rows with Zero or Negative UnitPrice:", (df["UnitPrice"] <= 0).sum())
```

```python
#Drop rows where Quantity or UnitPrice are zero or negative.
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]
```

```python
# Calculate total revenue for each transaction
df['TotalRevenue'] = df['Quantity'] * df['UnitPrice'] 
```

```python
# Show shape of the dataset after dropping null values
print("\nShape of the dataset after dropping null values and duplicates:", df.shape)
```

```python
# Show first 5 rows of the dataset after cleaning
print(df.head())
```

```python
# Export the cleaned dataset to a new Excel file
df.to_excel('Online Retail Cleaned.xlsx', index=False)
```
